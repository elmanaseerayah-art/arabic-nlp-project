```python
import streamlit as st
import requests
import json
import pandas as pd
from openai import OpenAI
```

```python
df=pd.read_csv('AAFAQ_Final_Cleaned.csv')
examples = df[['QuestionText','Answer']].head(5)
```

```python
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key="YOUR_API_KEY",
)
```

```python
st.title("Machine Translation: Arabic to English")
for index, row in examples.iterrows():
    arabic_question = row['QuestionText']
    arabic_answer = row['Answer']

    st.subheader(f"Example {index+1}")
    st.write("**Question (AR):**", arabic_question)
    st.write("**Answer (AR):**", arabic_answer)

    text_to_translate = f"Translate the following Arabic text to English:\nQuestion: {arabic_question}\nAnswer: {arabic_answer}"

    response = client.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are a helpful translator."},
            {"role": "user", "content": text_to_translate}
        ],
        max_tokens=500,
        temperature=0.3
    )

    translated_text = response.choices[0].message.content


    st.write("**Translation (EN):**", translated_text)
    st.write("---")
```
